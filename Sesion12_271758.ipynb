_Tu respuesta:_  
Los identificadores únicos carecen de poder predictivo porque funcionan únicamente como etiquetas de registro y no representan características intrínsecas del sujeto. Incluirlos en el modelo —incluso codificados— provoca un sobreajuste (overfitting), ya que el algoritmo memoriza casos específicos en lugar de aprender reglas generales. Como consecuencia, el sistema es incapaz de procesar datos nuevos (nuevos IDs) porque nunca los ha visto. El verdadero valor analítico reside en los patrones de comportamiento y atributos reales, tales como la edad, el historial de pagos o la frecuencia de consumo.

**3.3 — Agrupación "Top 10 + Otros"**

En clase agrupaste `country` de Netflix Titles en sus 10 categorías más frecuentes + `'Otros'`, reduciendo su cardinalidad.
```python
# Aplica el agrupamiento Top 10 + Otros sobre la columna de mayor cardinalidad
top10 = df['customerID'].value_counts().head(10)
categorias_top10 = top10.index.tolist()

df['customerID'] = df['customerID'].where(df['customerID'].isin(categorias_top10), 'Otros')

df['customerID'].value_counts()
```

```python
# 1. Buscamos los valores que NO se pueden convertir a número
valores_conflictivos = df[pd.to_numeric(df['TotalCharges'], errors='coerce').isna()]

# 2. Mostramos los valores únicos que están rompiendo la columna
print("Valores de texto o espacios encontrados:")
print(valores_conflictivos['TotalCharges'].unique())

# 3. Vemos cuántas filas tienen este problema
print(f"\nCantidad de filas con error: {len(valores_conflictivos)}")
```

_Tu respuesta:_ 11 columnas tienen espacios en blanco, debido a eso, esta columna fue determinada como objeto.

```python
# 4.2 — Corrige el tipo de TotalCharges.
# Pista: pd.to_numeric() con el parámetro errors= te puede ayudar a identificar
# o manejar los valores problemáticos que encontraste en 4.1.

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'].dtype
```
